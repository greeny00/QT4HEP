# 01 · Data Inspection

**Goal:** inspect the internal structure of `JetNtuple_RunIISummer16_13TeV_MC_1.h5`.
Which variables are stored, how they are stored (separate datasets or a single table), and how many jets the file contains.

Based on this output, the next step reads pT, η, ϕ, m and plots their distributions.

**How to run:** select the `qt4hep` kernel (top right) → **Run All**.

```python
from pathlib import Path
import h5py
import numpy as np

# Notebooks run from the project root (.vscode/settings.json: jupyter.notebookFileRoot)
DATA_PATH = Path("Data/JetNtuple_RunIISummer16_13TeV_MC_1.h5")

print("File exists:", DATA_PATH.exists())
print(f"Size: {DATA_PATH.stat().st_size / 1e6:.1f} MB")
print("h5py:", h5py.__version__, "| numpy:", np.__version__)
```

## 1. File structure

HDF5 files work like a folder tree: **groups** (folders) and **datasets** (arrays).
The cell below lists all of them with their shape and data type.

```python
def describe(name, obj):
    if isinstance(obj, h5py.Dataset):
        print(f"[dataset] {name:45s} shape={obj.shape}  dtype={obj.dtype}")
    else:
        print(f"[group]   {name}")

with h5py.File(DATA_PATH, "r") as f:
    print("Top-level keys:", list(f.keys()))
    print("-" * 80)
    f.visititems(describe)
```

## 2. Attributes

Some HDF5 files store column names or the table layout as attributes (e.g. when written with pandas).

```python
with h5py.File(DATA_PATH, "r") as f:
    def show_attrs(name, obj):
        if len(obj.attrs) > 0:
            print(f"{name}:")
            for k, v in list(obj.attrs.items())[:10]:
                print(f"    {k} = {str(v)[:120]}")
    print("Root attributes:", dict(list(f.attrs.items())[:10]))
    f.visititems(show_attrs)
```
